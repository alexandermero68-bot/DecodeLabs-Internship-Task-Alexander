# Project 1 — Data Cleaning & Preparation

**Input:** `Dataset for Data Analytics.xlsx`

**Output:** 'Dataset_Cleaned.xlsx'

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("Dataset for Data Analytics.xlsx")
df.head()
```

```python
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
df.info()
```

```python
df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.dtypes
```

```python
if "Date" in df.columns:
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
    print("Invalid/missing dates:", df["Date"].isna().sum())
```

```python
df = df.drop_duplicates().copy()
print("Shape after duplicate removal:", df.shape)
```

```python
print("Final missing values:")
print(df.isnull().sum())

print("Final duplicate rows:",
df.duplicated().sum())
```

```python
df.to_csv("Dataset_Cleaned.xlsx", index=False)
print("Saved to Dataset_Cleaned.xlsx")
```
